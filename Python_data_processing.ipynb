```python
records = [
    {"name": "Arun",    "dept": "IT",      "salary": 45000, "age": 24},
    {"name": "Priya",   "dept": "HR",      "salary": 38000, "age": 27},
    {"name": "Karthik", "dept": "IT",      "salary": 52000, "age": 31},
    {"name": "Divya",   "dept": "Finance", "salary": 41000, "age": 29},
    {"name": "Rahul",   "dept": "IT",      "salary": 60000, "age": 35},
    {"name": "Meena",   "dept": "HR",      "salary": 36000, "age": 26},
]

def get_values(data, key):
    return [r[key] for r in data]

def summary(data, key):
    values = get_values(data, key)
    return {
        "total": sum(values),
        "average": round(sum(values) / len(values), 2),
        "highest": max(values),
        "lowest": min(values),
    }

def group_by(data, key):
    groups = {}
    for r in data:
        groups.setdefault(r[key], []).append(r)
    return groups

def process_record(r):
    level = "Senior" if r["salary"] >= 50000 else "Junior"
    return {**r, "level": level, "annual_salary": r["salary"] * 12}

processed = [process_record(r) for r in records]

print("Salary summary:", summary(records, "salary"))
print("Age summary   :", summary(records, "age"))

print("\nDepartment-wise average salary:")
for dept, members in group_by(records, "dept").items():
    avg = sum(m["salary"] for m in members) / len(members)
    print(f"  {dept}: {avg:.2f} ({len(members)} employees)")

print("\nProcessed records (examples):")
for p in processed[:3]:
    print(p)
```
